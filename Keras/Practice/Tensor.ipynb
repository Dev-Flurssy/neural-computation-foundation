```python
# import libraries

import numpy as np
import tensorflow as tf
```

```python
# numpy to a tensor conversion

np_array = np.array([1, 2, 3, 4, 5])
print(np_array)
print()

np_to_tensor = tf.convert_to_tensor(np_array)
print(np_to_tensor)
```

```python
# another numpy to a tensor conversion method

np_array = np.array([1, 2, 3, 4, 5])
print(np_array)
print()

np_to_tensor = tf.constant(np_array)
print(np_to_tensor)
```

```python
# tensor to numpy conversion 

np_array = np.array([1, 2, 3, 4, 5])
print(np_array)
print()

np_to_tensor = tf.constant(np_array)
print(np_to_tensor)
print()

tensor_to_np = np_to_tensor.numpy()
print(tensor_to_np)
```

```python
# creation of constant and variable tensors

const_tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
var_tensor = tf.Variable([[7, 8, 9], [3, 5, 7]])

print(const_tensor)
print(const_tensor)
```

```python
# element wise addition of tensors

sum_tensor = tf.add(const_tensor, var_tensor)
print(sum_tensor)
```

```python
# element wise mathematical operation of tensors

sum_tensor = tf.add(const_tensor, var_tensor)
print(sum_tensor)
print()

diff_tensor = tf.subtract(const_tensor, var_tensor)
print(diff_tensor)
print()

mul_tensor = tf.multiply(const_tensor, var_tensor)
print(mul_tensor)
print()

div_tensor = tf.divide(var_tensor, const_tensor)
print(div_tensor)
```

```python
# reshaping of tensor

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
reshaped_tensor = tf.reshape(tensor, [3, 2])
print(reshaped_tensor )
```

```python
# flattening of tensor

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
flattened_tensor = tf.reshape(tensor, [-1])
print(flattened_tensor)
```

```python
# add dimension to a tensor

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
expanded_tensor = tf.expand_dims(tensor, axis=0)
print(expanded_tensor)
```

```python
# reduce dimension of a tensor

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
reduced_tensor = tf.reduce_sum(tensor, axis=1, keepdims=True)
print(reduced_tensor)
```

```python
# remove dimension of a tensor

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
reduced_tensor = tf.reduce_sum(tensor, axis=1, keepdims=True)
removed_tensor = tf.squeeze(reduced_tensor)
print(removed_tensor)
```

```python
# add sequence tensor

tensor = tf.constant([[9, 0, 7], [4, 5, 6]])
sequence_tensor = tf.reshape(tensor, [1, 2, 3])
print(sequence_tensor )
```

```python
# add transpose tensor

tensor = tf.constant([[9, 0, 7], [4, 5, 6]])
transpose_tensor = tf.transpose(tensor, perm=[1, 0])
print(transpose_tensor)
```

```python
# add complex tensor

tensor = tf.constant([[9, 0, 7], [4, 5, 6]])
complex_tensor = tf.reshape(tensor, [3, -1, 1])
print(complex_tensor)
```

```python
# batch matrix multiplication

batch_one = tf.random.normal([64, 10, 20])
batch_two = tf.random.normal([64, 20, 30])

batch_multiple = tf.matmul(batch_one, batch_two)
print(batch_multiple)
```

```python
# using distribution strategy

strategy = tf.distribute.MirroredStrategy()

with strategy.scope():
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(512, activation='relu'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```
