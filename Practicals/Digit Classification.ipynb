```python
import numpy as np
```

```python
class Network:
    def __init__(self, sizes):
        self.num_layers = len(sizes)
        self.sizes = sizes

        self.biases = []

        for layer_size in sizes[1:]:
            self.biases.append(np.random.randn(layer_size, 1))

        self.weights = []

        for i in range(len(sizes) - 1):
            input_neurons = sizes[i] 
            output_neurons = sizes[i + 1]

            self.weights.append(
                np.random.randn(output_neurons, input_neurons)
            )
```
